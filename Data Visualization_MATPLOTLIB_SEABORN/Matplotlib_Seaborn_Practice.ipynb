### Q5 — Bar, Horizontal Bar and Stacked Bar

**Dataset**
```python
departments = ["IT","HR","Sales","Finance"]
male = [42,18,35,25]
female = [28,32,25,20]
```

**Tasks**
1. Create a grouped bar chart comparing male and female employees.
2. Create a horizontal bar chart of total employees by department.
3. Create a stacked bar chart showing gender composition.
4. Add legends and data labels where useful.
5. State one practical use of each bar-chart form.

````python
departments = ["IT","HR","Sales","Finance"]
male = [42,18,35,25]
female = [28,32,25,20]
total = [m + f for m, f in zip(male, female)]
x = np.arange(len(departments))
width = 0.35

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Grouped bar chart
axes[0].bar(x - width/2, male, width, label='Male', color='steelblue')
axes[0].bar(x + width/2, female, width, label='Female', color='salmon')
axes[0].set_xticks(x)
axes[0].set_xticklabels(departments)
axes[0].set_title("Grouped Bar: Gender by Department")
axes[0].legend()

# Horizontal bar chart
axes[1].barh(departments, total, color='mediumseagreen')
for i, v in enumerate(total):
    axes[1].text(v + 1, i, str(v), va='center')
axes[1].set_title("Total Employees by Department")
axes[1].set_xlabel("Total Employees")

# Stacked bar chart
axes[2].bar(departments, male, label='Male', color='steelblue')
axes[2].bar(departments, female, bottom=male, label='Female', color='salmon')
axes[2].set_title("Stacked Bar: Gender Composition")
axes[2].legend()

plt.tight_layout()
plt.show()

print('''Practical uses:
- Grouped bar : best for direct side-by-side comparison of two categories per group.
- Horizontal bar : useful when category labels are long or there are many categories.
- Stacked bar : useful for showing part-to-whole composition alongside totals.''')
````

````python
marks = [45,52,55,61,64,65,67,70,71,72,74,75,76,78,80,82,85,88,90,95,38,42]
department_count = [35,25,20,20]
departments = ["CSE","IT","ECE","ME"]

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

axes[0].hist(marks, bins=8, color='cornflowerblue', edgecolor='black')
axes[0].set_title("Histogram of Marks")
axes[0].set_xlabel("Marks")
axes[0].set_ylabel("Frequency")

axes[1].pie(department_count, labels=departments, autopct='%1.1f%%',
            colors=sns.color_palette("pastel"))
axes[1].set_title("Department-wise Student Share")

axes[2].boxplot(marks, vert=True)
axes[2].set_title("Box Plot of Marks")
axes[2].set_ylabel("Marks")

plt.tight_layout()
plt.show()

q1, q3 = np.percentile(marks, [25, 75])
iqr = q3 - q1
lower, upper = q1 - 1.5*iqr, q3 + 1.5*iqr
outliers = [m for m in marks if m < lower or m > upper]
print("Possible outliers (via IQR rule):", outliers if outliers else "None detected")
print("Observation: The histogram is most useful for understanding the overall shape/spread "
      "of the marks distribution, while the box plot is best for quickly spotting outliers and quartiles.")
````

2. Create a single grouped bar chart showing all three metrics.

````python
models = ["Model A","Model B","Model C","Model D"]
accuracy = [0.82,0.87,0.84,0.91]
precision = [0.80,0.85,0.83,0.89]
recall = [0.78,0.88,0.81,0.90]

# 3-row subplot layout
fig, axes = plt.subplots(3, 1, figsize=(8, 10), sharex=True)
metrics = [("Accuracy", accuracy, 'steelblue'),
           ("Precision", precision, 'darkorange'),
           ("Recall", recall, 'seagreen')]
for ax, (name, vals, color) in zip(axes, metrics):
    ax.bar(models, vals, color=color)
    ax.set_title(name)
    ax.set_ylim(0.7, 1.0)
plt.tight_layout()
plt.show()

# Single grouped bar chart
x = np.arange(len(models))
width = 0.25
fig, ax = plt.subplots(figsize=(9, 5))
ax.bar(x - width, accuracy, width, label='Accuracy', color='steelblue')
ax.bar(x, precision, width, label='Precision', color='darkorange')
ax.bar(x + width, recall, width, label='Recall', color='seagreen')
ax.set_xticks(x)
ax.set_xticklabels(models)
ax.set_title("Model Metrics Comparison (Grouped Bar)")
ax.legend()
plt.tight_layout()
plt.show()

print('''Comparison: The grouped bar chart is more readable for direct model-to-model comparison
since all three metrics for a model sit next to each other; the 3-row layout is better for
seeing the trend of a single metric across models.

Observations:
1. Model D has the highest accuracy, precision and recall among all models.
2. Model B outperforms Model C on all three metrics despite being alphabetically earlier.''')
````

### Q13 — Object-Oriented API + Twin Axis

**Dataset**
```python
months = ["Jan","Feb","Mar","Apr","May","Jun"]
revenue = [10,12,14,13,17,20]
customers = [100,115,130,125,155,180]
```

**Tasks**
1. Build the plot using the object-oriented API.
2. Put revenue on the left y-axis and customers on a right y-axis using twinx().
3. Clearly label both axes.
4. Use a combined legend.
5. Explain one advantage and one risk of using two y-axes.

````python
months = ["Jan","Feb","Mar","Apr","May","Jun"]
revenue = [10,12,14,13,17,20]
customers = [100,115,130,125,155,180]

fig, ax1 = plt.subplots(figsize=(9, 5.5))
l1, = ax1.plot(months, revenue, color='tab:blue', marker='o', label='Revenue ($M)')
ax1.set_xlabel("Month")
ax1.set_ylabel("Revenue ($M)", color='tab:blue')
ax1.tick_params(axis='y', labelcolor='tab:blue')

ax2 = ax1.twinx()
l2, = ax2.plot(months, customers, color='tab:red', marker='s', label='Customers')
ax2.set_ylabel("Customers", color='tab:red')
ax2.tick_params(axis='y', labelcolor='tab:red')

lines = [l1, l2]
ax1.legend(lines, [l.get_label() for l in lines], loc='upper left')
plt.title("Revenue and Customers Over Time")
plt.tight_layout()
plt.show()

print('''Advantage: Two y-axes let you compare trends of two differently-scaled metrics on one chart
without one flattening the other.
Risk: Dual axes can visually mislead by implying correlation/alignment between series that
may not truly be related in that way.''')
````

````python
values = [1,2,3,4,5,7,10,15,25,40,80,150,300,700,1500]
idx = range(len(values))

fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))

axes[0].plot(idx, values, marker='o', color='teal')
axes[0].set_title("Linear Scale")
axes[0].set_xlabel("Index")
axes[0].set_ylabel("Value")
axes[0].grid(alpha=0.3)

axes[1].plot(idx, values, marker='o', color='darkorange')
axes[1].set_yscale('log')
axes[1].set_title("Logarithmic Scale (y-axis)")
axes[1].set_xlabel("Index")
axes[1].set_ylabel("Value (log scale)")
axes[1].grid(True, which='both', alpha=0.3)

plt.tight_layout()
plt.show()

print('''Comparison: On the linear scale, small early values are squashed near zero and hard to read,
while the log scale spreads them out and makes proportional (multiplicative) growth easier to see.

Log scaling is useful in data science when data spans several orders of magnitude
(e.g. income, population, website traffic) or when relative/percentage change matters more
than absolute change.''')
````

### Q16 — Seaborn Themes and Context

**Dataset**
```python
months = ["Jan","Feb","Mar","Apr","May","Jun"]
sales = [120,145,138,160,175,190]
profit = [20,25,22,31,35,42]
```

**Tasks**
1. Create a Seaborn line plot for sales and profit.
2. Experiment with at least two Seaborn styles.
3. Experiment with context settings appropriate for a report or presentation.
4. Use a readable color palette.

````python
months = ["Jan","Feb","Mar","Apr","May","Jun"]
sales = [120,145,138,160,175,190]
profit = [20,25,22,31,35,42]
df16 = pd.DataFrame({"Month": months, "Sales": sales, "Profit": profit})
df16_long = df16.melt(id_vars="Month", var_name="Metric", value_name="Value")

for style in ["whitegrid", "darkgrid"]:
    sns.set_style(style)
    sns.set_context("talk" if style == "whitegrid" else "notebook")
    plt.figure(figsize=(8, 5))
    sns.lineplot(data=df16_long, x="Month", y="Value", hue="Metric",
                 marker='o', palette="Set2")
    plt.title(f"Sales & Profit — style='{style}'")
    plt.tight_layout()
    plt.show()

sns.set_style("whitegrid")  # reset

print('''Seaborn differs from raw Matplotlib by providing attractive statistical defaults
(themes, palettes, confidence intervals) out of the box, and by working directly with
tidy Pandas DataFrames plus built-in statistical aggregation (e.g. automatic mean/CI in
lineplot/barplot), whereas Matplotlib requires manual styling and statistics.''')
````

### Q22 — Box Plot and Outlier Detection

**Dataset**
```python
department = ["IT"]*10 + ["HR"]*10
salary = [45000,48000,50000,52000,55000,57000,59000,60000,62000,120000,
          40000,42000,43000,45000,47000,48000,50000,51000,53000,55000]
```

**Tasks**
1. Create a boxplot of salary by department.
2. Identify possible outliers visually.
3. Compare median and spread between departments.
4. Explain what Q1, median, Q3 and IQR represent.
5. Do not remove any value; only identify and discuss it.

````python
department = ["IT"]*10 + ["HR"]*10
salary = [45000,48000,50000,52000,55000,57000,59000,60000,62000,120000,
          40000,42000,43000,45000,47000,48000,50000,51000,53000,55000]
df22 = pd.DataFrame({"Department": department, "Salary": salary})

plt.figure(figsize=(8, 5.5))
sns.boxplot(data=df22, x="Department", y="Salary", hue="Department",
            palette="Set3", legend=False)
plt.title("Salary Distribution by Department")
plt.tight_layout()
plt.show()

for dept in df22["Department"].unique():
    vals = df22[df22["Department"] == dept]["Salary"]
    q1, med, q3 = np.percentile(vals, [25, 50, 75])
    iqr = q3 - q1
    outliers = vals[(vals < q1 - 1.5*iqr) | (vals > q3 + 1.5*iqr)].tolist()
    print(f"{dept}: Q1={q1}, Median={med}, Q3={q3}, IQR={iqr}, Outliers={outliers}")

print('''
Q1 = 25th percentile (value below which 25% of data falls)
Median = 50th percentile (middle value)
Q3 = 75th percentile (value below which 75% of data falls)
IQR = Q3 - Q1, the spread of the middle 50% of data, used to flag outliers.
IT has a visible high outlier (120000) pulling its spread and mean upward compared to HR.''')
````
