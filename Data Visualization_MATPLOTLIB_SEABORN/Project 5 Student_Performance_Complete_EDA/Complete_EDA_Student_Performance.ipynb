### Complete EDA: Student Performance

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

**Dataset**
```python
data = {

    "Attendance":[65,70,75,80,85,90,95,60,72,82,88,98,77,84,93],

    "FinalScore":[48,55,62,68,75,82,90,40,58,70,85,94,64,78,91]
}
```

**Tasks**
1. Perform a complete visual EDA.
2. Plot distributions of all numerical features.
3. Create box plots to inspect outliers.
4. Create a correlation heatmap.
5. Create scatter/regression plots for StudyHours vs FinalScore and Attendance vs FinalScore.
6. Create a pairplot for the main numerical features.
7. Identify at least five meaningful observations.
8. State which variables appear visually associated with FinalScore, without claiming causation.

````python
data34 = {
    "StudyHours":[2,3,4,5,6,7,8,1,3,5,7,9,4,6,8],
    "Attendance":[65,70,75,80,85,90,95,60,72,82,88,98,77,84,93],
    "SleepHours":[5,6,6,7,7,8,8,5,6,7,7,8,6,7,8],
    "Assignments":[4,5,6,7,8,9,10,3,5,7,9,10,6,8,10],
    "FinalScore":[48,55,62,68,75,82,90,40,58,70,85,94,64,78,91]
}
df34 = pd.DataFrame(data34)

# 1 & 2. Distributions of all numerical features
fig, axes = plt.subplots(2, 3, figsize=(16, 9))
for ax, col in zip(axes.flatten(), df34.columns):
    sns.histplot(df34[col], kde=True, color='steelblue', ax=ax)
    ax.set_title(f"Distribution: {col}")
axes.flatten()[-1].axis('off')
plt.tight_layout()
plt.show()

# 3. Box plots for outliers
fig, axes = plt.subplots(1, 5, figsize=(18, 4.5))
for ax, col in zip(axes, df34.columns):
    sns.boxplot(y=df34[col], color='lightcoral', ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()

# 4. Correlation heatmap
plt.figure(figsize=(7, 6))
sns.heatmap(df34.corr(), annot=True, fmt=".2f", cmap="coolwarm", center=0, square=True)
plt.title("Correlation Heatmap: Student Performance")
plt.tight_layout()
plt.show()

# 5. Scatter/regression plots
fig, axes = plt.subplots(1, 2, figsize=(13, 5.5))
sns.regplot(data=df34, x="StudyHours", y="FinalScore", color='teal', ax=axes[0])
axes[0].set_title("StudyHours vs FinalScore")
sns.regplot(data=df34, x="Attendance", y="FinalScore", color='darkorange', ax=axes[1])
axes[1].set_title("Attendance vs FinalScore")
plt.tight_layout()
plt.show()

# 6. Pairplot
sns.pairplot(df34, diag_kind='kde')
plt.suptitle("Pairplot: Student Performance Features", y=1.02)
plt.show()

print('''Observations:
1. StudyHours and FinalScore show a very strong positive linear relationship.
2. Attendance is also strongly positively correlated with FinalScore.
3. SleepHours has a positive but comparatively weaker association with FinalScore.
4. No extreme outliers are visible in any of the boxplots; distributions are fairly smooth.
5. Assignments completed closely tracks StudyHours, suggesting more consistent students
   both study more and complete more assignments.

StudyHours, Attendance and Assignments all appear visually associated with FinalScore;
this reflects association only, not proof that any one of them directly causes higher scores.''')
````
