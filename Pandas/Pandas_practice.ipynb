## Topic: `apply()`, `map()`, and `lambda`

**Difficulty:** Medium (Placement Level)

### Tasks

1. Create a new column `Bonus` that is **10%** of `Salary` using `apply()` with `lambda`.
2. Create a new column `Total_Salary` equal to `Salary + Bonus`.
3. Use `map()` to convert the `Department` values as follows:
   - `HR` → `Human Resources`
   - `IT` → `Information Technology`
   - `Finance` → `Finance Department`
   - `Sales` → `Sales Department`
4. Create a new column `Grade` based on `Salary`:
   - `Salary >= 65000` → `"A"`
   - `Salary >= 55000` → `"B"`
   - Otherwise → `"C"`
5. Use `apply()` to create a new column `Experience_Level`:
   - `Experience >= 5` → `"Senior"`
   - Otherwise → `"Junior"`
6. Display the final DataFrame.


```python
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie", "David", "Eva", "Frank"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Sales"],
    "Salary": [45000, 60000, 55000, 70000, 48000, 52000],
    "Experience": [2, 5, 3, 7, 1, 4]
})
df
```

```python
df['Bonus'] = df['Salary'].apply(lambda x: x * 0.10)
df['Total_Salary'] = df['Salary'] + df['Bonus']
dept_mapping = {
"HR":"Human Resources",
"IT":"Information Technology",
"Finance": "Finance Department",
"Sales": "Sales Department"
}
df['Department'] = df['Department'].map(dept_mapping)
df['Grade'] = df['Salary'].apply(lambda x: "A" if x >= 65000 else ("B" if x >= 55000 else "C"))
df['Experience_Level'] = df['Experience'].apply(lambda x: "Senior" if x >= 5 else "Junior")
print(df)
```

# Pandas Challenge 24/100

## Topic: Window Functions

**Difficulty:** Hard (Placement Level)

### Tasks

1. Calculate the running total of `Sales` using `cumsum()`.
2. Calculate the running maximum of `Sales`.
3. Calculate the running minimum of `Sales`.
4. Calculate a **3-day moving average** of `Sales`.
5. Calculate a **3-day moving sum** of `Sales`.
6. Calculate the expanding average of `Sales`.
7. Display the final DataFrame with all new columns.

```python
df = pd.DataFrame({
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "Sales": [100, 120, 90, 150, 130, 170, 160]
})
# Task 1: Running total
df["Running_Total"] = df["Sales"].cumsum()

# Task 2: Running maximum
df["Running_Max"] = df["Sales"].cummax()

# Task 3: Running minimum
df["Running_Min"] = df["Sales"].cummin()

# Task 4: 3-day moving average
df["Moving_Avg_3"] = df["Sales"].rolling(window=3).mean()

# Task 5: 3-day moving sum
df["Moving_Sum_3"] = df["Sales"].rolling(window=3).sum()

# Task 6: Expanding average
df["Expanding_Avg"] = df["Sales"].expanding().mean()

# Task 7: Display final DataFrame
print(df)
```


# Pandas Challenge 25/100

## Topic: Feature Engineering (Final Challenge)

**Difficulty:** Hard (Placement Level)

### Tasks

1. Create an `Age_Group` column using `cut()` with these bins:
   - 20–30 → `"Young"`
   - 30–40 → `"Mid"`
   - 40–50 → `"Senior"`
2. Create a `Salary_Level` column using `qcut()` with **3 equal-frequency groups** labeled `"Low"`, `"Medium"`, and `"High"`.
3. Create a `Performance_Bonus` column:
   - If `Salary >= 60000`, bonus is **15%** of Salary.
   - Otherwise, bonus is **5%** of Salary.
4. Create a `Salary_per_Experience` column equal to `Salary / Experience`.
5. Create a `High_Performer` column:
   - `"Yes"` if `Salary >= 60000` **and** `Experience >= 5`.
   - Otherwise `"No"`.
6. Create a `Capped_Salary` column by limiting Salary to the range **45000–65000** using `clip()`.
7. Display the final DataFrame.


```python
df = pd.DataFrame({
    "Name": ["Alice","Bob","Charlie","David","Eva","Frank","Grace","Henry"],
    "Age": [22,35,28,45,24,38,42,30],
    "Salary": [45000,60000,55000,70000,48000,52000,65000,75000],
    "Experience": [2,5,3,7,1,4,8,6]
})
# Task 1: Age_Group using cut()
df["Age_Group"] = pd.cut(
    df["Age"],
    bins=[20, 30, 40, 50],
    labels=["Young", "Mid", "Senior"],
    include_lowest=True
)

# Task 2: Salary_Level using qcut()
df["Salary_Level"] = pd.qcut(
    df["Salary"],
    q=3,
    labels=["Low", "Medium", "High"]
)

# Task 3: Performance_Bonus
df["Performance_Bonus"] = df["Salary"].apply(
    lambda x: x * 0.15 if x >= 60000 else x * 0.05
)

# Task 4: Salary_per_Experience
df["Salary_per_Experience"] = df["Salary"] / df["Experience"]

# Task 5: High_Performer
df["High_Performer"] = df.apply(
    lambda row: "Yes"
    if row["Salary"] >= 60000 and row["Experience"] >= 5
    else "No",
    axis=1
)

# Task 6: Capped_Salary
df["Capped_Salary"] = df["Salary"].clip(45000, 65000)

# Task 7: Display final DataFrame
print(df)
```
